```python
import pandas as pd
from pathlib import Path
```

```python
data_path = Path("../data/energydata_complete.csv")

df = pd.read_csv(data_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
print("Shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nFirst 5 rows:")
display(df.head())
```

```python
missing_values = df.isnull().sum()

print("Missing values per column:")
display(missing_values.sort_values(ascending=False))

print("\nTotal missing values:", missing_values.sum())
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
print("Are rv1 and rv2 identical?")
print(df["rv1"].equals(df["rv2"]))

print("\nCorrelation between rv1 and rv2:")
print(df["rv1"].corr(df["rv2"]))

print("\nNumber of rows where values differ:")
print((df["rv1"] != df["rv2"]).sum())
```

```python
df.describe().T
```

```python
df["date"] = pd.to_datetime(df["date"], errors="raise")

print("Date data type:", df["date"].dtype)
print("Earliest timestamp:", df["date"].min())
print("Latest timestamp:", df["date"].max())
print("Missing timestamps:", df["date"].isna().sum())
```

```python
print("Data types:")
display(df.dtypes.value_counts())

print("\nNon-numeric columns:")
print(df.select_dtypes(exclude="number").columns.tolist())

print("\nNumeric columns:")
print(df.select_dtypes(include="number").columns.tolist())
```

```python
time_differences = df["date"].sort_values().diff().dropna()

print("Most common time intervals:")
display(time_differences.value_counts().head())

print("\nNumber of non-10-minute intervals:",
      (time_differences != pd.Timedelta(minutes=10)).sum())

print("\nDuplicate timestamps:", df["date"].duplicated().sum())
```

```python
print(
    "Already in chronological order:",
    df["date"].is_monotonic_increasing
)
```

```python
df_clean = df.copy()

df_clean = df_clean.drop(columns=["rv2"])

print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)
print("rv1 retained:", "rv1" in df_clean.columns)
print("rv2 retained:", "rv2" in df_clean.columns)
print("Missing values:", df_clean.isna().sum().sum())
print("Duplicate rows:", df_clean.duplicated().sum())
```

```python
from pathlib import Path

processed_dir = Path("../data/processed")
processed_dir.mkdir(parents=True, exist_ok=True)

output_path = processed_dir / "energydata_processed.csv"
df_clean.to_csv(output_path, index=False)

print("Saved to:", output_path)
print("Saved shape:", df_clean.shape)
```

```python
df_check = pd.read_csv(
    output_path,
    parse_dates=["date"]
)

print("Reloaded shape:", df_check.shape)
print("Date dtype:", df_check["date"].dtype)
print("Missing values:", df_check.isna().sum().sum())
print("Duplicate rows:", df_check.duplicated().sum())
print("rv2 present:", "rv2" in df_check.columns)

assert df_check.shape == (19735, 28)
assert df_check["date"].dtype.kind == "M"
assert df_check.isna().sum().sum() == 0
assert "rv2" not in df_check.columns

print("\nAll verification checks passed!")
```

- No missing-value imputation was required.
- No duplicate-row removal was required.

### 6. Processed Dataset
- Created a cleaned copy named `df_clean`.

- Verified the saved file by reloading it with `parse_dates=["date"]`.

- Final verification: zero missing values, zero duplicate rows, and `rv2` absent.

### 7. Key Learning
Preprocessing decisions should be based on data inspection and project requirements rather than applying transformations automatically. Transformations that learn parameters, such as scaling or imputation, must be fitted using training data only to prevent data leakage.
