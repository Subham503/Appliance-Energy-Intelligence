# 04 - Feature Engineering

## Objective

Create and investigate meaningful features from the
preprocessed appliance energy dataset to potentially
improve the predictive capability of future ML models.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/processed/energydata_processed.csv")

df.head()
```

```python
# Inspect dataset information
df.info()
```

```python
# Inspect all column names
df.columns.tolist()
```

```python
# Convert date column to datetime
df['date'] = pd.to_datetime(df['date'])

# Verify the conversion
df['date'].dtype
```

```python
df[['date']].head()
```

```python
# Extract time-based features

df['hour'] = df['date'].dt.hour
df['day'] = df['date'].dt.day
df['month'] = df['date'].dt.month
df['day_of_week'] = df['date'].dt.day_of_week
```

```python
df[['date', 'hour', 'day', 'month', 'day_of_week']].head(15)
```

```python
# Create weekend indicator
df['is_weekend'] = (df['day_of_week'] >= 5).astype(int)
```

```python
df[['date', 'day_of_week', 'is_weekend']].head(20)
```

```python
df[df['is_weekend'] == 1][
    ['date', 'day_of_week', 'is_weekend']
].head(10)
```

```python
df['is_weekend'].value_counts()
```

```python
# Cyclical encoding of hour

df['hour_sin'] = np.sin(2 * np.pi * df['hour'] / 24)
df['hour_cos'] = np.cos(2 * np.pi * df['hour'] / 24)
```

```python
df[['hour', 'hour_sin', 'hour_cos']].head(15)
```

```python
df[df['hour'].isin([0, 1, 22, 23])][
    ['hour', 'hour_sin', 'hour_cos']
].drop_duplicates().sort_values('hour')
```

```python
# Calculate indoor-outdoor temperature difference
df['temp_diff_indoor_outdoor'] = df['T1'] - df['T_out']
```

```python
df[['T1', 'T_out', 'temp_diff_indoor_outdoor']].head(10)
```

```python
df['temp_diff_indoor_outdoor'].corr(df['Appliances'])
```

```python
# Calculate indoor-outdoor humidity difference
df['humidity_diff_indoor_outdoor'] = df['RH_1'] - df['RH_out']
```

```python
df[['RH_1', 'RH_out', 'humidity_diff_indoor_outdoor']].head(10)
```

```python
df['humidity_diff_indoor_outdoor'].corr(df['Appliances'])
```

```python
temp_cols = [f'T{i}' for i in range(1, 10)]
```

```python
df['avg_indoor_temp'] = df[temp_cols].mean(axis=1)
```

```python
df[['T1', 'T2', 'T3', 'T4', 'T5',
    'T6', 'T7', 'T8', 'T9',
    'avg_indoor_temp']].head()
```

```python
df['avg_indoor_temp'].corr(df['Appliances'])
```

```python
hum_cols = [f'RH_{i}' for i in range(1, 10)]
```

```python
hum_cols
```

```python
df['avg_indoor_humidity'] = df[hum_cols].mean(axis=1)
```

```python
df[['RH_1', 'RH_2', 'RH_3', 'RH_4', 'RH_5', 'RH_6', 'RH_7', 'RH_8', 'RH_9',
    'avg_indoor_humidity']].head()
```

```python
df['avg_indoor_humidity'].corr(df['Appliances'])
```

```python
engineered_cols = [
    'hour', 'day', 'month', 'day_of_week', 'is_weekend',
    'hour_sin', 'hour_cos',
    'temp_diff_indoor_outdoor',
    'humidity_diff_indoor_outdoor',
    'avg_indoor_temp',
    'avg_indoor_humidity'
]

df[engineered_cols + ['Appliances']].corr()['Appliances'].sort_values(
    ascending=False
)
```

## Summary

In this notebook, feature engineering was performed on the preprocessed appliance energy dataset.

### Features Created

- **Time-based features:** hour, day, month, day_of_week, is_weekend.
- **Cyclical features:** hour_sin, hour_cos.
- **Domain-based features:** temp_diff_indoor_outdoor, humidity_diff_indoor_outdoor.
- **Aggregation features:** avg_indoor_temp, avg_indoor_humidity.

### Key Observations

- Created 11 new features.
- Individual correlations with Appliances were generally weak.
- Humidity difference showed a positive correlation (~0.182).
- Hour showed the strongest positive correlation (~0.217) among the engineered features.

These correlations are exploratory and do not establish causation or predictive usefulness.

### Validation

- Final dataset shape: 19,735 rows × 39 columns.
- Missing values: 0.
- Duplicate rows: 0.
- Infinite values in engineered features: 0.
- Original features were preserved.
